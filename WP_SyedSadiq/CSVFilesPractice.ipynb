```python
%sql
--Create the delta table with cleaned column names
CREATE TABLE winequality_red AS
SELECT 
    `fixed acidity` AS fixed_acidity,
    `volatile acidity` AS volatile_acidity,
    `citric acid` AS citric_acid,
    `residual sugar` AS residual_sugar,
    chlorides,
    `free sulfur dioxide` AS free_sulfur_dioxide,
    `total sulfur dioxide` AS total_sulfur_dioxide,
    density,
    pH,
    sulphates,
    alcohol,
    quality
FROM read_files('/Volumes/dbacademy/default/tutorials/winequality-red.csv',
                format => 'csv', 
                header => true, 
                delimiter => ';');
```

```python
%sql
--Preview the delta table
SELECT * FROM winequality_red limit 10;
```

```python
%sql
DESCRIBE TABLE EXTENDED dbacademy.default.winequality_red;
```

```python
# 1. Read the csv files from the volume into the spark dataframe
from pyspark.sql.functions import current_timestamp, from_utc_timestamp, date_format

df = spark.read.format("csv") \
    .option("header", "true") \
    .option("delimiter", ";") \
    .option("inferSchema", "true") \
    .load("/Volumes/dbacademy/default/tutorials/winequality-red.csv")
df = df.toDF(*[col.replace(" ", "_") for col in df.columns])
df = df.withColumn("batch_run_date", date_format(from_utc_timestamp(current_timestamp(), "Asia/Kolkata"), "dd-MM-yyyy HH:mm:ss"))
df.printSchema()
display(df)
df.write.mode("append").saveAsTable("dbacademy.default.winequality_red")
```

```python
%sql
ALTER TABLE winequality_red_missing_col SET TBLPROPERTIES ('delta.columnMapping.mode' = 'name');
```

```python
%sql

COPY INTO winequality_red_missing_col
FROM (
  SELECT *,
    date_format(from_utc_timestamp(current_timestamp(), 'Asia/Kolkata'), 'dd-MM-yyyy HH:mm:ss') AS batch_run_date
  FROM '/Volumes/dbacademy/default/tutorials'
)
FILEFORMAT = CSV
FORMAT_OPTIONS ('delimiter' = ';', 'header' = 'true')
COPY_OPTIONS ('mergeSchema' = 'true'); --, "force" = "true")
```
