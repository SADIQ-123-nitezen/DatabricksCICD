```sql
use dbacademy.jobschedule;
```

```sql
COPY INTO dbacademy.jobschedule.tbl_job_data
FROM (
  SELECT 
    `fixed acidity`::double AS fixed_acidity,
    `volatile acidity`::double AS volatile_acidity,
    `citric acid`::double AS citric_acid,
    `residual sugar`::double AS residual_sugar,
    chlorides::double,
    `free sulfur dioxide`::double AS free_sulfur_dioxide,
    `total sulfur dioxide`::double AS total_sulfur_dioxide,
    density::double,
    pH::double,
    sulphates::double,
    alcohol::double,
    quality::int,
    date_format(from_utc_timestamp(current_timestamp(), 'Asia/Kolkata'), 'dd-MM-yyyy HH:mm:ss') AS batch_run_date,
    _metadata.file_path,
    _metadata.file_modification_time
  FROM '/Volumes/dbacademy/jobschedule/data_ingestion'
)
FILEFORMAT = CSV
FORMAT_OPTIONS ('header' = 'true', 'delimiter' = ';')
```

```sql
select a.batch_run_date, a.file_path, a.file_modification_time, count(*) data_count from dbacademy.jobschedule.tbl_job_data a group by a.batch_run_date, a.file_path , a.file_modification_time;
```
